```python
import QuantLib as ql

# Parameters
s = 100.0        # Underlying asset price
k = 100.0        # Strike price
t = 1.0          # Time to expiration in years
r = 0.05         # Risk-free interest rate
sigma = 0.2      # Volatility
div_yield = 0.0  # Dividend yield

# Set up the valuation date
today = ql.Date.todaysDate()
ql.Settings.instance().evaluationDate = today

# Choose a calendar and day count convention
calendar = ql.UnitedStates(ql.UnitedStates.NYSE)
day_count = ql.Actual365Fixed()

# Calculate the maturity date
maturity_date = calendar.advance(today, ql.Period(int(t * 365 + 0.5), ql.Days))

# Set up the market data
spot = ql.SimpleQuote(s)
volatility = ql.BlackConstantVol(today, calendar, sigma, day_count)
dividend_yield = ql.FlatForward(today, div_yield, day_count)
risk_free_rate = ql.FlatForward(today, r, day_count)

# Set up the option
payoff = ql.PlainVanillaPayoff(ql.Option.Put, k)
exercise = ql.AmericanExercise(today, maturity_date)
option = ql.VanillaOption(payoff, exercise)

# Set up the pricing engine
process = ql.BlackScholesMertonProcess(
    ql.QuoteHandle(spot),
    ql.YieldTermStructureHandle(dividend_yield),
    ql.YieldTermStructureHandle(risk_free_rate),
    ql.BlackVolTermStructureHandle(volatility),
)

steps = 100      # Adjusted to match Rust implementation
engine = ql.BinomialVanillaEngine(process, "crr", steps)
option.setPricingEngine(engine)

# Calculate delta
delta = option.delta()
print(f"American Put Delta (QuantLib): {delta}")
```
