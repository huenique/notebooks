```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Parameters for the simulation
np.random.seed(42)  # For reproducibility
num_days = 182  # Total days for simulation
funding_updates_per_day = 3  # Funding updates every 8 hours
T = num_days * funding_updates_per_day  # Total steps (182 days * 3 updates per day)
dt = 1 / funding_updates_per_day  # Time increment (for each 8-hour step)
mu = 0.05  # Drift (annual return)
sigma = 0.2  # Volatility (annual)
S0 = 50000  # Initial price for BTC spot
initial_basis = 0.0001  # Initial basis rate (small initial basis)
funding_threshold = 0.0005  # Arbitrage threshold for entering positions
slippage_tolerance = 0.001  # Slippage tolerance for closing positions (lowered)
stop_loss_threshold = -0.0005  # Stop-loss if basis widens too much (more relaxed)
take_profit_threshold = 0.0001  # Take-profit threshold
investment_size = 10000  # Amount invested in each position (in USD)
max_basis_diff = 10  # Maximum allowed difference between spot and perpetual prices in USD

# Brownian motion for spot prices
def generate_brownian_motion(T, dt, mu, sigma, S0):
    """Simulate Brownian motion with drift."""
    W = np.random.standard_normal(size=T)  # Random normal values
    W = np.cumsum(W) * np.sqrt(dt)  # Brownian motion path
    time_series = np.linspace(0, T * dt, T)
    S = S0 * np.exp((mu - 0.5 * sigma**2) * time_series + sigma * W)
    return S

# Simulate prices and basis
def simulate_prices_and_basis(T, S0, initial_basis, mu, sigma, max_basis_diff):
    spot_prices = generate_brownian_motion(T, dt, mu, sigma, S0)
    perp_prices = np.copy(spot_prices)  # Initialize perp prices
    for t in range(1, T):
        perp_prices[t] = spot_prices[t] + np.random.uniform(-max_basis_diff, max_basis_diff)  # Keep perp prices close to spot
    basis = (perp_prices - spot_prices) / spot_prices  # Calculate basis as percentage
    return spot_prices, perp_prices, basis

# Simulate arbitrage returns
def simulate_arbitrage_strategy(T, S0, funding_threshold, slippage_tolerance, stop_loss_threshold, take_profit_threshold, investment_size, max_basis_diff):
    spot_prices, perp_prices, basis = simulate_prices_and_basis(T, S0, initial_basis, mu, sigma, max_basis_diff)
    
    # Track performance metrics
    returns = np.zeros(T)
    positions = np.zeros(T)  # Track positions: -1 for long perp, 1 for short perp, 0 for no position
    pnl = 0  # Total profit/loss
    active_position = 0  # Track active positions
    entry_basis_rate = 0  # Store basis rate at entry
    total_pnl = 0  # Track total cumulative PnL
    
    print(f"{'Time':<5} {'Spot Price':<15} {'Perp Price':<15} {'Basis (USD)':<15} {'Basis Rate':<15}")
    print("-" * 60)

    for t in range(1, T):
        price_diff = perp_prices[t] - spot_prices[t]  # In USD
        basis_rate = price_diff / spot_prices[t]

        # Print the updates in spot price, perp price, basis (price difference), and basis rate
        print(f"{t:<5} {spot_prices[t]:<15.2f} {perp_prices[t]:<15.2f} {price_diff:<15.2f} {basis_rate:<15.4f}")

        # If no active position, enter based on basis rate thresholds
        if active_position == 0:
            if basis_rate > funding_threshold:
                active_position = 1  # Short perp, long spot (positive funding rate arbitrage)
                entry_basis_rate = basis_rate
                print(f"Time {t}: Entered short perp, long spot position (basis rate: {basis_rate:.4f})")
            elif basis_rate < -funding_threshold:
                active_position = -1  # Long perp, short spot (negative funding rate arbitrage)
                entry_basis_rate = basis_rate
                print(f"Time {t}: Entered long perp, short spot position (basis rate: {basis_rate:.4f})")

        # If in a position, calculate PnL based on the funding rate and the notional value
        if active_position == 1:
            # We are short perp, long spot, so we earn funding when the basis narrows
            pnl = investment_size * (entry_basis_rate - basis_rate)
            
            if basis_rate < take_profit_threshold:
                # Take profit when basis narrows
                total_pnl += pnl - (investment_size * slippage_tolerance)  # Adjust for slippage
                active_position = 0
                print(f"Time {t}: Took profit on short perp, long spot position (Total PnL: {total_pnl:.2f}, basis rate: {basis_rate:.4f})")
            elif basis_rate < stop_loss_threshold:
                # Stop-loss if the basis widens too much in the wrong direction
                total_pnl -= pnl
                active_position = 0
                print(f"Time {t}: Stopped loss on short perp, long spot position (Total PnL: {total_pnl:.2f}, basis rate: {basis_rate:.4f})")

        elif active_position == -1:
            # We are long perp, short spot, so we pay funding when the basis narrows
            pnl = investment_size * (basis_rate - entry_basis_rate)
            
            if basis_rate > -take_profit_threshold:
                # Take profit when basis narrows
                total_pnl += pnl - (investment_size * slippage_tolerance)  # Adjust for slippage
                active_position = 0
                print(f"Time {t}: Took profit on long perp, short spot position (Total PnL: {total_pnl:.2f}, basis rate: {basis_rate:.4f})")
            elif basis_rate > stop_loss_threshold:
                # Stop-loss if the basis widens too much in the wrong direction
                total_pnl -= pnl
                active_position = 0
                print(f"Time {t}: Stopped loss on long perp, short spot position (Total PnL: {total_pnl:.2f}, basis rate: {basis_rate:.4f})")

    return returns, positions, spot_prices, perp_prices, total_pnl

# Run the simulation
returns, positions, spot_prices, perp_prices, total_pnl = simulate_arbitrage_strategy(
    T, S0, funding_threshold, slippage_tolerance, stop_loss_threshold, take_profit_threshold, investment_size, max_basis_diff)

# Plotting the results
plt.figure(figsize=(10, 6))
plt.subplot(2, 1, 1)
plt.plot(spot_prices, label="Spot Price (BTC)")
plt.plot(perp_prices, label="Perp Price (BTC)")
plt.title("Simulated Spot and Perpetual Prices")
plt.legend()

plt.subplot(2, 1, 2)
plt.plot(positions, label="Arbitrage Position (-1 = Long Perp, 1 = Short Perp)")
plt.title("Arbitrage Position Over Time (8-hour intervals)")
plt.legend()

plt.tight_layout()
plt.show()

# Summarize results
print(f"Total PnL from strategy: {total_pnl:.2f} USD over {num_days} days ({T} 8-hour periods)")
```
